```python
import pandas as pd
```

```python
df = pd.read_excel("NovaTech_Sales_Raw_Data.xlsx")
```

```python
#Remove duplicates
df = df.drop_duplicates()
```

```python
type(df)
```

```python
df.columns = df.columns.str.strip()

for col in df.select_dtypes(include=['object', 'string']).columns:
    df[col] = df[col].str.strip()
```

```python
df.head()
```

```python
df.info()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['State'] = df['State'].fillna('Unknown')
df['Customer Segment'] = df['Customer Segment'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# Time features

df['Year'] = df['Order Date'].dt.year
df['Quarter'] = 'Q' + df['Order Date'].dt.quarter.astype(str)
df['Month'] = df['Order Date'].dt.month_name()
df['Month Number'] = df['Order Date'].dt.month
```

```python
df[['Order Date','Year','Quarter','Month','Month Number']].head()
```

```python
df['Profit Margin %'] = (
    df['Profit (INR)'] /
    df['Net Sales (INR)']
) * 100
```

```python
df[['Net Sales (INR)','Profit (INR)','Profit Margin %']].head()
```

```python
df['Profit Per Unit'] = (
    df['Profit (INR)'] /
    df['Quantity']
)
```

```python
df[['Quantity','Profit (INR)','Profit Per Unit']].head()
```

```python
df['Discount Category'] = pd.cut(
    df['Discount (%)'],
    bins=[-1,0,5,10,20],
    labels=[
        'No Discount',
        'Low Discount',
        'Medium Discount',
        'High Discount'
    ]
)
```

```python
df[['Discount (%)','Discount Category']].head(10)
```

```python
df['High Value Order'] = df['Net Sales (INR)'].apply(
    lambda x: 'Yes' if x >= 500000 else 'No'
)
```

```python
df[['Net Sales (INR)','High Value Order']].head(10)
```

```python
df[['Year',
    'Quarter',
    'Month',
    'Month Number',
    'Profit Margin %',
    'Profit Per Unit',
    'Discount Category',
    'High Value Order']].head()
```

```python
df.to_excel('NovaTech_Cleaned_Data.xlsx', index=False)

print("NovaTech_Cleaned_Data.xlsx saved successfully")
```
